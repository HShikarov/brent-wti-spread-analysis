# What Explains Movements in the Brent-WTI Crude Oil Spread?

**Data Science in Economics (BEE2041) Empirical Project**

**Executive Summary:**
The aim of the analysis is to examine the association of physical oil market fundamentals (Cushing crude stocks, refinery utilization, and crude imports) with the Brent-WTI price spread using weekly data from 2020 to 2026 through OLS regression and generated visualizations

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.formula.api as smf
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.stattools import durbin_watson

pd.set_option('display.max_columns', None)
sns.set_style("whitegrid")

BASE_DIR = Path.cwd()
RAW_DIR = BASE_DIR / "data" / "raw"
PROCESSED_DIR = BASE_DIR / "data" / "processed"
OUTPUT_DIR = BASE_DIR / "output"
FIGURES_DIR = OUTPUT_DIR / "figures"

prices_path = RAW_DIR / "prices_daily.csv"
eia_path = RAW_DIR / "eia_weekly.csv"

print("Working directory:", BASE_DIR)
print("Prices file exists:", prices_path.exists(), prices_path)
print("EIA file exists:", eia_path.exists(), eia_path)
print("Processed dir exists:", PROCESSED_DIR.exists(), PROCESSED_DIR)
print("Output dir exists:", OUTPUT_DIR.exists(), OUTPUT_DIR)
print("Figures dir exists:", FIGURES_DIR.exists(), FIGURES_DIR)
```

## 1. Introduction

The objective of the research is to analyze **to what extent are weekly movements in the Brent-WTI spread associated with U.S. physical oil market fundamentals, such as Cushing stocks, refinery utilization, and crude imports over the 2020 to 2026 period.**

On that note, the Brent and WTI crude oil price spread are the two primary benchmark crude oils that reflect the regional supply-demand; however, while they both demonstrate global benchmarks, Brent prices European crude while WTI indicates U.S. Gulft Coast pricing at Cushing, Oklahoma, and their price difference is a key indicator of dislocation between global crude markets.  Taking that into account, understanding this spread is crucial for physical trading, transport bottlenecks, and refinery economics - factors that fuel the world trade

```python
prices = pd.read_csv(prices_path)

print("Before cleaning prices:")
print(prices.shape)
print(prices.isna().sum())

prices.columns = [c.strip().lower() for c in prices.columns]

expected_price_cols = {"date", "wti_price", "brent_price"}
missing_price_cols = expected_price_cols - set(prices.columns)
if missing_price_cols:
    raise ValueError(f"Missing columns in prices_daily.csv: {missing_price_cols}")

prices = prices.dropna(how="all")
prices = prices.dropna(subset=["date"])

prices["date"] = pd.to_datetime(prices["date"], errors="coerce")
prices["wti_price"] = pd.to_numeric(prices["wti_price"], errors="coerce")
prices["brent_price"] = pd.to_numeric(prices["brent_price"], errors="coerce")

prices = prices.dropna(subset=["date"])
prices = prices.dropna(subset=["wti_price", "brent_price"])   # important change
prices = prices[prices["date"] >= "2020-01-01"].copy()
prices = prices.sort_values("date").drop_duplicates(subset=["date"])

prices.to_csv(prices_path, index=False)

print("\nAfter cleaning prices:")
print(prices.shape)
print(prices.isna().sum())
print(prices.head())
```

```python
eia = pd.read_csv(eia_path)

print("Before cleaning EIA:")
print(eia.shape)
print(eia.isna().sum())

eia.columns = [c.strip().lower() for c in eia.columns]

expected_eia_cols = {"date", "cushing_stocks", "refinery_util", "crude_imports"}
missing_eia_cols = expected_eia_cols - set(eia.columns)
if missing_eia_cols:
    raise ValueError(f"Missing columns in eia_weekly.csv: {missing_eia_cols}")

eia = eia.dropna(how="all")
eia = eia.dropna(subset=["date"])

eia["date"] = pd.to_datetime(eia["date"], errors="coerce")
eia["cushing_stocks"] = pd.to_numeric(eia["cushing_stocks"], errors="coerce")
eia["refinery_util"] = pd.to_numeric(eia["refinery_util"], errors="coerce")
eia["crude_imports"] = pd.to_numeric(eia["crude_imports"], errors="coerce")

eia = eia.dropna()
eia = eia[eia["date"] >= "2020-01-01"].copy()
eia = eia.sort_values("date").drop_duplicates(subset=["date"])

eia.to_csv(eia_path, index=False)

print("\nAfter cleaning EIA:")
print(eia.shape)
print(eia.isna().sum())
print(eia.head())
```

## 2. Data

The analysis uses two raw data files, 'prices_daily.csv' and 'eia_weekly.csv'. Given that, the inital input contains daily Brent and WTI spot prices, while the second contains weekly EIA (U.S. Energy Information Administration) petroleum indicators, respectively. However,  post-cleaning, the price data cover 1,546 daily observations.

Similarily, the sample period of the analysis is weekly data from 2020 to 2026, sourced from EIA spot prices and EIA weekly petroleum series; and, the following variables were used in the analysis:

* 'brent_price,' for Europe brent spot price in American dollars per barrel
* 'wti_price,' for Cushing, Oklahoma WTI spot price in American dollars per barrel
* 'spread,' for brent minus WTI
* 'cushing_stocks' crude oil stocks at Cushing, Oklahoma, in million barrels
* 'refinery_util,' refinery capacity utilization, in percentage
* 'crude_imports,' weekly U.S. crude imports
* The daily prices were aggregated to Friday ending (weekly) averages to match the weekly EIA series

Considering that, the petroleum market variables are available at weekly frequency, converting the daily price data into Friday-ending weekly averages would ensure that both datasets are on the same time unit. Hence, then by merging the two datasets into week-ending date and creating a one-period lag of the spread, the result would be a consistent weekly panel that can be used for the analysis.

```python
prices["spread"] = prices["brent_price"] - prices["wti_price"]

weekly_prices = prices.set_index("date").resample("W-FRI").mean(numeric_only=True).reset_index()
weekly_prices.rename(columns={"date": "week_ending"}, inplace=True)

df = pd.merge(
    weekly_prices,
    eia,
    left_on="week_ending",
    right_on="date",
    how="inner"
)

df = df.sort_values("week_ending")
df["spread_lag1"] = df["spread"].shift(1)
df = df.dropna().copy()

processed_path = PROCESSED_DIR / "analysis_data.csv"
df.to_csv(processed_path, index=False)

print("Merged shape:", df.shape)
print(df.head())
print(df[["spread", "cushing_stocks", "refinery_util", "crude_imports"]].describe())
```

```python
print("\nSummary Statistics:")
print(df[['spread', 'cushing_stocks', 'refinery_util', 'crude_imports']].describe().round(2))
```

### 2.1 Data Summary

In the end, the sample period covers 328 weekly observations from January 2020 to April 2026. The Brent-WTI spread averaged $4.23 per barrel with significant variation -- SD = $2.74 -- ranging from a minimum of $-2.97 to a maximum of $22.43. Cushing crude stocks averaged 33.29 million barrels, refinery utilization averaged 87.89%, and crude imports averaged 6250.65 thousand barrels per day. As a result, these summary values provide useful context for interpreting both the figures and the regression coefficients later on in the analysis


## 3. Method

To study the relationship between physical market conditions and the Brent–WTI spread, an ordinary least squares (OLS) regression is conducted, through:
$$
[ 
pread_t = \beta_0 + \beta_1 cushing\_stocks_t + \beta_2 refinery\_util_t + \beta_3 crude\_imports_t + \beta_4 spread_{t-1} + \varepsilon_t 
]
$$
The dependent variable is the weekly Brent–WTI spread, where the explanatory variables are Cushing crude inventories, refinery utilization, crude imports, and one period lag of the spread. Including the lagged spread is curcial, since benchmark price differentials are often persistent over time: especially considering that the market adjustments are gradual rather than immediate.

However, this specification is an **associational** model rather than a causal one -- taking into account that the aim is not to claim one variable definitively casuing the spread to move: but to examine whether these market fundamentals are related to weekly variation in the spread. Correspondingly, OLS is estimated via HC3 standard errors to account for potential heteroskedasticity in the analysis.

```python
formula = "spread ~ cushing_stocks + refinery_util + crude_imports + spread_lag1"
model = smf.ols(formula=formula, data=df).fit(cov_type="HC3")

print(model.summary())

dw_stat = durbin_watson(model.resid)
bp_test = het_breuschpagan(model.resid, model.model.exog)

print("\nDurbin-Watson:", round(dw_stat, 3))
print("Breusch-Pagan p-value:", round(bp_test[1], 3))

with open(OUTPUT_DIR / "regression_results.txt", "w") as f:
    f.write(model.summary().as_text())
```

## 4.1 Regression Results

The regression output illustrates that the Brent–WTI spread is highly persistent over time -- where the lagged spread (coefficient) is statistically significant at **0.8736**, indicating that there is a strong assoication between the previous and current week’s spread. Hence, this dominant result in the model suggests that the spread dynamics evolve gradually rather than being fully reset each week.

In contrast, the three physical market variables are not statistically significant, however. The coefficient on **Cushing stocks** is **-0.0089**, the coefficient on **refinery utilization** is **0.017**, and the coefficient on **crude imports** is slightly negative **-0.0002** -- but all three coefficients have their respective p-values above the conventional thresholds. This suggests that once persistence is controlled, the data does not provide strong statistical evidence for the variables independently explaining the weekly spread movements in this model.

On the other hand, the model has an **R-squared of 0.719**, demonstrating that the model does explain a large share of variation in the spread. Nevertheless, most of that explanatory power comes from the lagged dependent variable rather than from the physical fundamental coefficients. Similarily, the **Durbin–Watson statistic is 2.107** (a test to measure the relationship between consecutive observations, ranging from 0 to 4), which does not suggest severe residual autocorrelation. The **Breusch–Pagan p-value is 0.004**, since the p-value is below the appropriate threshold (p < 0.05), there is evidence to assume heteroskedasticity: which also further supports the use of robust standard errors in the analysis.

```python
# Figures 1 & 2: Price levels & Spread Dynamics

# Figure 1: Brent & WTI Over Time
fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(df["week_ending"], df["brent_price"], 
        label="Brent (Europe)", linewidth=2, color='#2E86AB')
ax.plot(df["week_ending"], df["wti_price"], 
        label="WTI (Cushing, OK)", linewidth=2, color='#A23B72')
ax.set_xlabel("Date", fontweight='bold')
ax.set_ylabel("Price ($/barrel)", fontweight='bold')
ax.set_title("Brent and WTI Crude Oil Prices (2020-2026)", 
             fontweight='bold', fontsize=14)
ax.legend(loc='upper left', frameon=True, shadow=True)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(FIGURES_DIR / "01_prices_over_time.png", dpi=300, bbox_inches='tight')
plt.show()

# Figure 2: Spread Over Time
fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(df["week_ending"], df["spread"], linewidth=2, color='#F18F01')
ax.axhline(0, color='gray', linestyle="--", linewidth=1, alpha=0.5)
ax.axhline(df["spread"].mean(), color='green', linestyle=":", 
           linewidth=1.5, alpha=0.8, label=f"Mean: ${df['spread'].mean():.2f}")

# Highlighting the COVID-19 & Russia-Ukraine Conflict periods
covid_start = pd.to_datetime('2020-03-01')
covid_end = pd.to_datetime('2020-05-31')
ax.axvspan(covid_start, covid_end, alpha=0.2, color='red', label='COVID-19 Crisis')

conflict_start = pd.to_datetime('2022-02-24')
conflict_end = pd.to_datetime('2022-04-07')
ax.axvspan(conflict_start, conflict_end, alpha=0.2, color='orchid', label='Russia-Ukraine War')

ax.set_xlabel("Date", fontweight='bold')
ax.set_ylabel("Spread ($/barrel)", fontweight='bold')
ax.set_title("Brent-WTI Spread Over Time", fontweight='bold', fontsize=14)
ax.legend(loc='upper left', frameon=True, shadow=True)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(FIGURES_DIR / "02_spread_over_time.png", dpi=300, bbox_inches='tight')
plt.show()
```

### 4.2 Price Levels and Spread Dynamics

The first two figures illustrate that Brent and WTI prices move closely together over time, which is consistent with the fact that both are benchmark crude oils are influenced by broad global energy market conditions: however, the gap between them is not constant. On that note, the spread varies substantially between the periods, highlighting that meaningful pricing changes between the crude markets over time.

Consequently, the spread varies across subperiods: in 2020, the series shows unusually sharp movements, reflecting extreme market disruption during the COVID period. Furthermore, the Russian-Ukrainian conflict in 2022 also contributed to the widening, and the sharp movements. Nevertheless, the spread post-2022 remains mostly positive despite certain fluctuations around its average rather than staying fixed at one level where the recent data also illustrate instances of renewed widening. Ultimately, this pattern suggests that the spread should be viewed as a dynamic relative-price indicator rather than as a stable trend.

In the end, the figures demonstrate an important distinction: the level of oil prices and the relative price between Brent and WTI are not the same thing. Considering that, Brent and WTI may rise or fall together, but the spread can still change considerably depending on regional balances, logistics, and market structure.

```python
# Figures 3 & 4: Inventories & Estimated Coefficients

# Figure 3: Stocks versus spread
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(df["cushing_stocks"], df["spread"], alpha=0.6)

z = np.polyfit(df["cushing_stocks"], df["spread"], 1)
p = np.poly1d(z)
x_sorted = np.sort(df["cushing_stocks"])
ax.plot(x_sorted, p(x_sorted), linestyle="--", linewidth=2)

corr = df["cushing_stocks"].corr(df["spread"])
ax.text(
    0.05, 0.95, f"Correlation = {corr:.3f}",
    transform=ax.transAxes,
    verticalalignment="top",
    bbox=dict(boxstyle="round", facecolor="white", alpha=0.8)
)

ax.set_xlabel("Cushing Stocks (million barrels)")
ax.set_ylabel("Brent-WTI Spread ($/barrel)")
ax.set_title("Cushing Stocks vs Brent-WTI Spread")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(FIGURES_DIR / "03_stocks_scatter.png", dpi=300)
plt.show()

# Figure 4: Coefficient plot
coef_df = pd.DataFrame({
    "variable": ["Cushing Stocks", "Refinery Utilization", "Crude Imports", "Lagged Spread"],
    "coefficient": [
        model.params["cushing_stocks"],
        model.params["refinery_util"],
        model.params["crude_imports"],
        model.params["spread_lag1"],
    ],
    "stderr": [
        model.bse["cushing_stocks"],
        model.bse["refinery_util"],
        model.bse["crude_imports"],
        model.bse["spread_lag1"],
    ]
})

fig, ax = plt.subplots(figsize=(10, 6))
for idx, row in coef_df.iterrows():
    ax.barh(idx, row["coefficient"], xerr=1.96 * row["stderr"], capsize=5)

ax.axvline(0, color="black", linewidth=1)
ax.set_yticks(range(len(coef_df)))
ax.set_yticklabels(coef_df["variable"])
ax.set_xlabel("Coefficient")
ax.set_title("Regression Coefficients with 95% Confidence Intervals")
ax.grid(True, alpha=0.3, axis="x")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "04_coefficients.png", dpi=300)
plt.show()
```

### 4.3 Inventories and Estimated Coefficients

The scatter plot between Cushing stocks and the Brent–WTI spread suggests a **moderately negative correlation** (R² = - 0.347). In the raw data, higher inventories at Cushing tend to be associated with a narrower or lower spread. This is economically plausible, since inventory accumulation at a U.S. delivery hub can reflect local oversupply and thereby put relative downward pressure on WTI-linked pricing.

However, the regression coefficient plot tells a more nuanced story. Once other controls are included—especially the lagged spread—the estimated inventory effect becomes much smaller in statistical terms and is no longer significant. The same is true for refinery utilization and crude imports. Their confidence intervals cross zero, which indicates that the model cannot reject the possibility of no independent association in this specification.

Similarly, the difference between the scatter plot and the multivariate model is important. Hence, a simple bivariate relationship captures common time trends -- on the other hand, the regression isolates the partial association of each variable holding the others fixed. In this analysis, however, that partial relationship appears to be much weaker than solely the raw scatter plot would indicate.

```python
# Figure 5: Refinery Utilization & Spread

fig, ax1 = plt.subplots(figsize=(12, 6))
ax1.plot(df["week_ending"], df["spread"], linewidth=2, label="Spread")
ax1.set_xlabel("Date")
ax1.set_ylabel("Spread ($/barrel)")
ax1.grid(True, alpha=0.3)

ax2 = ax1.twinx()
ax2.plot(df["week_ending"], df["refinery_util"], linewidth=2, alpha=0.7, label="Refinery Utilization")
ax2.set_ylabel("Refinery Utilization (%)")

ax1.set_title("Refinery Utilization and Brent-WTI Spread")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "05_refinery_spread.png", dpi=300)
plt.show()

print("All figures saved to:", FIGURES_DIR)
```

### 4.4 Economic Interpretation

The strongest interpretation in the data is the persistence of the Brent–WTI spread -- which suggests that relative crude pricing does not adjust instantaneously. Instead, once the spread widens or narrows, that movement tends to carry forward into the next week. This is consistent with the idea that transport constraints, storage conditions, contractual arrangements, and physical delivery systems can create gradual rather than immediate adjustment.

The insignificance of Cushing stocks, refinery utilization, and crude imports does not necessarily mean that physical fundamentals are irrelevant, but that in this particular case, their independent effect is difficult to detect after controlling for persistence. Hence, a possible deducation would be the possibility of these variables mattering in more complex ways that a linear model can measure. Another is that other omitted factors - such as infrastructure constraints, export capacity, geopolitical shocks, or sanctions, or changes in demand - play an important role in shaping the spread.

On that account, the results should be interpreted carefully -- in particular, since the model indicates that short-run spread movements are dominated by persistence, but it does not rule out a meaningful economic role for inventories, refining conditions, and trade flows in broader crude market adjustment.

## 5. Limitations

This project has several limitations: firstly, the analysis is more associational but than a causal, hence the coefficients should not be interpreted as causal effects; secondly, the regression includes only a minimal set of explanatory variables, meaning that certain omitted variables may be important. On that note, factors such as export flows, freight costs, pipeline capacity, broad geopolitical events, and shifts in global market structure could all influence the Brent–WTI spread.

Thirdly, the sample period from 2020 to 2026 includes highly unusual instances, especially the COVID shock and subsequent market adjustment and the Russian-Ukrainian 2022 conflict. These events may make the relationships in the data less stable than in a more normal period. Lastly, the use of weekly averages may be more suitable for short-run movements, and the presence of heteroskedasticity would suggest an uncertainty that varies across time, which is an indicator that the model is a rather simplified representation of a more complex case.

## 6. Conclusion

The analysis examined whether weekly movements in the Brent–WTI crude oil spread are associated with U.S. physical oil market fundamentals: where, in the end, three main findings emerge.

To begin with, the Brent and WTI prices move closely together over time, but the spread between them changes significantly across the period. Furthermore, Cushing inventories show a negative simple relationship with the spread in the raw data, and, lastly, once the regression controls for refinery utilization, crude imports, and especially the lagged spread, persistence becomes the dominant statistical feature of the data.

Overall, the results suggest that short-run Brent–WTI spread dynamics are highly persistent, while the independent role of the selected physical fundamentals is weaker in this simple weekly model. Thereby, a more elaborate specification could explore nonlinear effects, or alternative time horizons, but even so, this analysis highlights an important point: relative crude oil pricing is shaped not only by broad price levels, but by the evolving structure of regional market conditions

To conclude, the analysis conducted has practical relevance in effectively understanding crude oil price dynamics: the strong persistence in the spread suggests the presence of constrained physical infrastructure and transportation costs within oil abarbitrage opportunities, arising from WTI-Brent spread. Regarding traders and analysts, this implies that spread movements are, however, only partially predictable from past values, though the modest correlation (R² = | 0.347 |) indicates that other factors -- for instance geopolitical events and supply disruptions -- play important roles. On that note, future research could further incorporate these additional variables or examine whether the relationships vary across different market dynamics.
